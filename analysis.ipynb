# Student Study Habits Analysis

## Project Introduction

The purpose of this project is to analyze a university student dataset using Python. The project explores how different factors such as study habits, scholarship type, reading frequency, and exam preparation are associated with students' academic performance. Data analysis and visualization techniques are used to identify meaningful patterns and provide insights into student achievement.

## Project Objectives

- Explore the student dataset.
- Identify factors that influence academic performance.
- Create visualizations to understand patterns.
- Draw conclusions based on the analysis.

## Research Questions

1. Does weekly study time influence student grades?
2. Does reading frequency have an effect on academic performance?
3. Is scholarship type associated with student grades?
4. Does exam preparation improve academic performance?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("data/StudentsPerformance_with_headers.csv")

# Show first 5 rows
df.head()
```

```python
# Dataset information
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Distribution of Student Grades

plt.figure(figsize=(8,5))
sns.countplot(x="GRADE", data=df)

plt.title("Distribution of Student Grades")
plt.xlabel("Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Weekly Study Hours vs Grade

plt.figure(figsize=(10,5))
sns.boxplot(x="Weekly study hours", y="GRADE", data=df)

plt.title("Weekly Study Hours vs Student Grade")
plt.xlabel("Weekly Study Hours")
plt.ylabel("Grade")

plt.show()
```

## Analysis 1: Weekly Study Hours vs Grade

This chart shows the relationship between students' weekly study hours and their academic grades. Students who study more hours generally tend to achieve better grades, although there are a few exceptions. This suggests that study time has a positive impact on academic performance.

```python
# Scholarship Type vs Grade

plt.figure(figsize=(10,5))
sns.boxplot(x="Scholarship type", y="GRADE", data=df)

plt.title("Scholarship Type vs Student Grade")
plt.xlabel("Scholarship Type")
plt.ylabel("Grade")

plt.show()
```

```python
# Reading Frequency vs Grade

plt.figure(figsize=(10,5))
sns.boxplot(x="Reading frequency", y="GRADE", data=df)

plt.title("Reading Frequency vs Student Grade")
plt.xlabel("Reading Frequency")
plt.ylabel("Grade")
plt.show()
```

```python
# Preparation to Midterm Exams vs Grade

plt.figure(figsize=(10,5))
sns.boxplot(x="Preparation to midterm exams 1", y="GRADE", data=df)

plt.title("Preparation for Midterm Exams vs Grade")
plt.xlabel("Preparation Type")
plt.ylabel("Grade")
plt.show()
```

```python
# Student Age Distribution

plt.figure(figsize=(8,5))
sns.histplot(df["Student Age"], bins=10, kde=True)

plt.title("Distribution of Student Age")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Correlation Heatmap

plt.figure(figsize=(14,10))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(), cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
